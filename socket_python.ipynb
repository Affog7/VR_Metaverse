```python
import asyncio
import websockets
import json

# Fonction de gestion des connexions
async def handler(websocket, path):
    print("Client connecté !")
    async for message in websocket:
        print("Message reçu :", message)

        # Traitement simple de la commande
        action = "none"
        if "avance" in message:
            action = "forward"
        elif "droite" in message and "tourne" in message:
            action = "rotate_right"
        elif "gauche" in message and "tourne" in message:
            action = "rotate_left"

        response = json.dumps({"action": action})
        await websocket.send(response)

# Démarrage du serveur
async def start_server():
    server = await websockets.serve(handler, "localhost", 8768)
    print("✅ Serveur WebSocket lancé sur ws://localhost:8768")
    return server

# Appel pour démarrer le serveur dans Jupyter
server = await start_server()
```

# Test du socket

```python
import asyncio
import websockets

async def test_client():
    uri = "ws://localhost:8768"
    async with websockets.connect(uri) as websocket:
        message = "tourne à droite"
        print("🟢 Envoi :", message)
        await websocket.send(message)

        response = await websocket.recv()
        print("📥 Réponse du serveur :", response)

await test_client()
```

# Arrêter le socket

```python
server.close()
await server.wait_closed()
print("🔴 Serveur WebSocket arrêté.")
```
